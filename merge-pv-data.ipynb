```python
import pandas as pd
import glob
import os
```

```python
# Path to folder containing all your solar site CSVs
csv_files = glob.glob("data/Dataset/Time series dataset/PV generation dataset/PV stations with panel level optimizer/Site level dataset/*.csv")

df_list = []
for file in csv_files:
    df = pd.read_csv(file)
    # Use filename (without extension) as a label column
    site_name = os.path.splitext(os.path.basename(file))[0]
    df["site"] = site_name
    df_list.append(df)

merged_df = pd.concat(df_list, ignore_index=True)
merged_df.to_csv("data/stage-1/merged_solar_data.csv", index=False)
```
